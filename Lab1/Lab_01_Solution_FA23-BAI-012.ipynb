## Task 06: String Cleaning Methods
**Topic:** String Methods – lower, upper, strip, split, join, replace

### (a) strip(), lower(), replace() applied step by step

```python
text = "   MACHINE Learning is AMAZING!!!   "

# (i) strip() to remove leading and trailing spaces
step1 = text.strip()
print("After strip():", repr(step1))

# (ii) lower() to normalise case
step2 = step1.lower()
print("After lower():", step2)

# (iii) replace() to remove all "!" characters
step3 = step2.replace("!", "")
print("After replace():", step3)
```

### (b) split(), list comprehension with upper(), and join()

```python
# Use the cleaned text from part (a)
cleaned_text = step3

# Tokenise into a list of words using split()
words_list = cleaned_text.split()
print("Word list:", words_list)

# Use upper() on each word inside a list comprehension
uppercase_list = [w.upper() for w in words_list]
print("Uppercase list:", uppercase_list)

# Join the uppercase words using " | " as separator
joined_string = " | ".join(uppercase_list)
print("Joined string:", joined_string)
```

### (d) normalize() function to clean raw text

```python
def normalize(text):
    """Strip whitespace, lowercase text, and remove commas and full-stops."""
    text = text.strip()
    text = text.lower()
    text = text.replace(",", "")
    text = text.replace(".", "")
    return text

# Test on two messy strings
sample1 = "  Hello, World.  "
sample2 = "NLP is Amazing, and Powerful.  "

print("Cleaned sample1:", normalize(sample1))
print("Cleaned sample2:", normalize(sample2))
```

### (b) Check alphabetic, numeric, or mixed tokens

```python
tokens = ["hello", "123", "world", "456abc", "python", "789", "NLP3"]

for token in tokens:
    if token.isalpha():
        print(f'"{token}" -> purely alphabetic')
    elif token.isdigit():
        print(f'"{token}" -> purely numeric')
    else:
        print(f'"{token}" -> mixed')
```

### (c) Separate words into three groups

```python
words = ["running", "jumped", "NLP", "ai2025", "data", "42"]

pure_words = []
pure_numbers = []
mixed_tokens = []

for w in words:
    if w.isalpha():
        pure_words.append(w)
    elif w.isdigit():
        pure_numbers.append(w)
    else:
        mixed_tokens.append(w)

print("Pure words:", pure_words)
print("Pure numbers:", pure_numbers)
print("Mixed tokens:", mixed_tokens)
```

### (d) validate_token() function

```python
def validate_token(token):
    """Return WORD, NUMBER, or MIXED depending on the token's content."""
    if token.isalpha():
        return "WORD"
    elif token.isdigit():
        return "NUMBER"
    else:
        return "MIXED"

test_tokens = ["chat", "99", "gpt4", "bert", "2024", "nlp!"]

for token in test_tokens:
    label = validate_token(token)
    print(f'{token} -> {label}')
```

## Task 08: Functions and Lambda for NLP Pipelines
**Topic:** Functions, Lambda – Reusable Text Processing

### (a) Regular function vs lambda function for cleaning text

```python
def clean(text):
    """Strip whitespace and convert text to lowercase."""
    return text.strip().lower()

# Equivalent lambda function
clean_lambda = lambda text: text.strip().lower()

sample_text = "   HELLO Natural Language Processing   "

print("clean():", clean(sample_text))
print("clean_lambda():", clean_lambda(sample_text))
```

### (b) tokenize() and remove_short() chained together

```python
def tokenize(text):
    """Split a cleaned string into a list of word tokens."""
    return text.split()

def remove_short(tokens, n):
    """Keep only tokens whose length is greater than n."""
    return [t for t in tokens if len(t) > n]

sentence3 = "I love working with NLP tools in Python daily"

# Chain both functions
tokens_result = tokenize(sentence3)
final_tokens = remove_short(tokens_result, 3)
print("Tokens after removing short words:", final_tokens)
```

### (c) word_lengths() function

```python
def word_lengths(tokens):
    """Return a dictionary mapping each word to its length."""
    return {word: len(word) for word in tokens}

sample_tokens = ["language", "model", "ai", "processing", "deep"]
lengths_dict = word_lengths(sample_tokens)
print("Word lengths:", lengths_dict)
```

### (d) Full nlp_pipeline() function combining previous functions

```python
def nlp_pipeline(text, min_len):
    """Clean text, tokenise it, remove short words, and return word lengths."""
    cleaned = clean(text)
    tokens = tokenize(cleaned)
    filtered_tokens = remove_short(tokens, min_len)
    return word_lengths(filtered_tokens)

# Test the pipeline on two different sentences
result1 = nlp_pipeline("Natural Language Processing is a growing field", 3)
result2 = nlp_pipeline("Deep learning models require a lot of data", 2)

print("Pipeline result 1:", result1)
print("Pipeline result 2:", result2)
```

## Task 09: map(), filter(), sorted() and Counter
**Topic:** map(), filter(), sorted(), Lambda, collections.Counter

### (a) map() with lambda for lowercase and word length

```python
words = ["Python", "NLP", "is", "a", "powerful", "tool", "for", "text"]

# (i) Convert every word to lowercase using map()
lower_words = list(map(lambda w: w.lower(), words))
print("Lowercase words:", lower_words)

# (ii) Compute the character length of each word using map()
word_lengths_list = list(map(lambda w: len(w), words))
print("Word lengths:", word_lengths_list)
```

### (b) filter() to create two filtered lists

```python
# (i) Words with more than 3 characters
words_long = list(filter(lambda w: len(w) > 3, words))
print("Words longer than 3 characters:", words_long)

# (ii) Words containing the letter "t" (case-insensitive)
words_with_t = list(filter(lambda w: "t" in w.lower(), words))
print("Words containing 't':", words_with_t)
```

### (c) sorted() in three different ways

```python
# (i) Alphabetically (default order)
alpha_sorted = sorted(words)
print("Alphabetical order:", alpha_sorted)

# (ii) By word length ascending
length_sorted_asc = sorted(words, key=len)
print("Length ascending:", length_sorted_asc)

# (iii) By word length descending
length_sorted_desc = sorted(words, key=len, reverse=True)
print("Length descending:", length_sorted_desc)
```

### (d) Counter from collections

```python
from collections import Counter

sentence4 = "nlp nlp python ai nlp python text ai nlp python text"
word_list = sentence4.split()

# Create a Counter object
freq = Counter(word_list)
print("Full counter:", freq)

# (i) 3 most common words
print("3 most common words:", freq.most_common(3))

# (ii) Frequency of the word "python"
print("Frequency of 'python':", freq["python"])

# (iii) Total number of words and size of unique vocabulary
print("Total number of words:", len(word_list))
print("Unique vocabulary size:", len(freq))
```
