## Task 1: Regex Fundamentals (Character Classes & Quantifiers)

Given a sample text with mixed content (words, numbers, symbols), write regex patterns for:
- a) Extract all digits from the text
- b) Extract all words that start with a capital letter
- c) Extract all sequences of exactly 4 digits (e.g., years)
- d) Extract all words that are between 3 to 6 characters long

```python
text = "In 2023, Ahmed paid Rs. 4500 for 3 books on 15 Jan 2024. Sara and Ali visited Lahore in 1999 & 2005! Python is fun, easy and very useful. Room #42 costs $120."
print(text)
```

**a)** `\d` matches one digit, so `findall` returns every single digit.

```python
digits = re.findall(r"\d", text)
print(digits)
```

**b)** `\b` = word boundary, `[A-Z]` = capital first letter, `\w*` = the rest of the word.

```python
capital_words = re.findall(r"\b[A-Z]\w*\b", text)
print(capital_words)
```

**c)** `\d{4}` = exactly 4 digits. The `\b` on both sides stops it from matching 4 digits inside a longer number.

```python
four_digits = re.findall(r"\b\d{4}\b", text)
print(four_digits)
```

**d)** `[A-Za-z]{3,6}` = 3 to 6 letters; `\b` on both sides makes sure the *whole* word has that length.

```python
words_3_to_6 = re.findall(r"\b[A-Za-z]{3,6}\b", text)
print(words_3_to_6)
```

## Task 2: Pattern-Based Extraction (Real World Data)

Given a paragraph containing emails, phone numbers, order IDs, dates, and percentages mixed together:
- a) Extract all email addresses  b) Extract all phone numbers (03XX-XXXXXXX)  c) Extract all dates in DD/MM/YYYY format  d) Extract all percentage values (e.g., 45%, 12.5%)

```python
paragraph = ("Order ORD-1021 was placed by ali.khan@gmail.com on 12/03/2024. "
             "Call 0300-1234567 or 0345-7654321 for support, or write to support@shop.com.pk. "
             "Order ORD-1022 on 25/12/2024 got a discount of 12.5% and a tax of 17%. "
             "Sara (sara_99@yahoo.com, 0321-5558888) ordered ORD-1023 on 01/01/2025 with 45% off.")
print(paragraph)
```

**a) Emails:** name part, `@`, domain, then one or more `.something` parts.

```python
emails = re.findall(r"[\w.+-]+@[\w-]+(?:\.[\w-]+)+", paragraph)
print(emails)
```

**b) Phone numbers:** `03`, two more digits, a dash, then exactly 7 digits.

```python
phones = re.findall(r"\b03\d{2}-\d{7}\b", paragraph)
print(phones)
```

**c) Dates:** `DD/MM/YYYY` = 2 digits / 2 digits / 4 digits.

```python
dates = re.findall(r"\b\d{2}/\d{2}/\d{4}\b", paragraph)
print(dates)
```

**d) Percentages:** digits, optionally a decimal part, then `%`.

```python
percentages = re.findall(r"\d+(?:\.\d+)?%", paragraph)
print(percentages)
```

## Task 3: Word Tokenization (NLTK vs spaCy)

a) Tokenize with NLTK's `word_tokenize()`  b) Tokenize with a blank spaCy pipeline  c) Write 2+ differences in a comment  d) Count tokens for each.

```python
para = ("I don't like waiting in long queues. It's a waste of time! "
        "My well-known friend said \"we can't wait\" and left. "
        "They're sure it'll be fine, isn't it?")
print(para)
```

**a) NLTK.** `word_tokenize` needs the `punkt_tab` data (downloaded once).  
*Note: if the download is blocked (e.g. no internet), the cell falls back to NLTK's own `NLTKWordTokenizer` applied sentence by sentence, which is what `word_tokenize` does internally.*

```python
from nltk.tokenize import word_tokenize, NLTKWordTokenizer

try:
    nltk.download("punkt_tab", quiet=True)
    nltk_tokens = word_tokenize(para)
except LookupError:
    # fallback: split into sentences with a simple rule, then use NLTK's word tokenizer
    nltk_tokens = []
    for sent in re.split(r"(?<=[.!?])\s+(?=[A-Z])", para):
        nltk_tokens += NLTKWordTokenizer().tokenize(sent)

print(nltk_tokens)
```

**b) spaCy** (blank English pipeline = only the tokenizer, no trained model).

```python
nlp_blank = spacy.blank("en")
spacy_doc = nlp_blank(para)
spacy_tokens = [token.text for token in spacy_doc]
print(spacy_tokens)
```

**d) Token counts**

```python
print("NLTK  tokens:", len(nltk_tokens))
print("spaCy tokens:", len(spacy_tokens))
```

**c) Differences** (see the comment in the cell below)

```python
# DIFFERENCES between NLTK and spaCy (2+ required):
# 1) Hyphenated words: NLTK keeps "well-known" as ONE token, spaCy splits it into
#    three tokens: "well", "-", "known".
# 2) Quotation marks: NLTK changes the double quote " into `` (opening) and ''
#    (closing), so the text of the token is changed. spaCy keeps the original " as is.
# 3) Because of these rules spaCy produces more tokens than NLTK (see the counts above).
# 4) Contractions are handled the same way by both: "don't" -> "do" + "n't",
#    "It's" -> "It" + "'s", "can't" -> "ca" + "n't".
```

## Task 4: Token Attributes (Categorizing Tokens)

Using a sentence with normal words, numbers, an email and a currency symbol:
a) Load it into a spaCy `Doc`  b) print `is_alpha`  c) print `like_num` and `like_email`  d) list tokens where `is_currency` is True.

```python
sentence = "Ali paid $50 for 3 books and emailed ali@example.com about 2 refunds."
doc = nlp_blank(sentence)       # a) the sentence is now a spaCy Doc
print([t.text for t in doc])
```

**b) `is_alpha`** – True only if the token is made of letters.

```python
for token in doc:
    print(f"{token.text:<18} is_alpha = {token.is_alpha}")
```

**c) `like_num` and `like_email`**

```python
for token in doc:
    print(f"{token.text:<18} like_num = {str(token.like_num):<6} like_email = {token.like_email}")
```

**d) Tokens where `is_currency` is True**

```python
currency_tokens = []
for token in doc:
    if token.is_currency:
        currency_tokens.append(token.text)
print(currency_tokens)
```
